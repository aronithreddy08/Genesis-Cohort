```sql

-- Q1. Find duplicate records
SELECT column_name, COUNT(*)
FROM table_name
GROUP BY column_name
HAVING COUNT(*) > 1;
```

```sql

-- Q2. Find the second highest salary
SELECT MAX(Salary) AS SecondHighestSalary
FROM Employee
WHERE Salary < (SELECT MAX(Salary) FROM Employee);
```

```sql

-- Q3. Find employees without a department
SELECT e.EmployeeID, e.EmployeeName
FROM Employee e
LEFT JOIN Department d
ON e.DepartmentID = d.DepartmentID
WHERE d.DepartmentID IS NULL;
```

```sql

-- Q4. Calculate total revenue per product
SELECT ProductID,
SUM(Quantity * Price) AS TotalRevenue
FROM Sales
GROUP BY ProductID;
     
```

```sql

-- Q5. Get the top 3 highest-paid employees
SELECT EmployeeName, Salary
FROM Employee
ORDER BY Salary DESC
LIMIT 3;
```

```sql
-- Q6. Customers who made purchases but never returned products
SELECT DISTINCT CustomerID
FROM Orders
WHERE CustomerID NOT IN (
    SELECT CustomerID
    FROM Returns
);
```
