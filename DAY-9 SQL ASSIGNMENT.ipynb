```sql
DROP TABLE IF EXISTS gold_sales;
CREATE TABLE gold_sales (
    Order_ID INT,
    Customer_ID INT,
    Customer_Name VARCHAR(50),
    Department VARCHAR(30),
    Order_Date DATE,
    Sales_Amount DECIMAL(10,2)
);
```

```sql
-- 1. Calculate total sales for each Department using SUM() OVER().
SELECT *,
       SUM(Sales_Amount) OVER(PARTITION BY Department) AS Total_Sales
FROM gold_sales;
```

```sql
-- 2. Calculate average sales for each Department using AVG() OVER().
select *,
    avg(Sales_Amount)over (partition by Department) As Avg_Sales
    From gold_sales;
```

```sql
-- 3. Calculate company-wide total sales using SUM() OVER().

select *,
    SUM(Sales_Amount)over () AS Total_Sales
    from gold_sales;
```

```sql
-- 4. Calculate company-wide average sales using AVG() OVER().
select *,
    Avg(Sales_Amount)over() AS Avg_Sales
    from  gold_sales;
```

```sql
-- 5. Calculate a running total of sales ordered by Order_Date.
select * ,
    sum(Sales_Amount)over(order by Order_Date) As Running_Sum
    from gold_sales;

    
```

```sql
-- 6. Calculate a running total separately for each Department.
select *,
    Sum(Sales_Amount) over(
        Partition by Department
        order by Order_Date
    ) As Departments_runSum
    from gold_sales;
```

```sql
-- 7. Find the highest-selling customer in each Department using ROW_NUMBER().

SELECT *
FROM (
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY Department
               ORDER BY Sales_Amount DESC
           ) AS rn
    FROM gold_sales
) t
WHERE rn = 1;
```

```sql
-- 8. Find the top 2 sales levels in each Department using DENSE_RANK().
Select *
from(
    select *,
            DENSE_RANK() over (
            Partition by Department
            Order by Sales_Amount Desc
        )As RN
    From gold_sales
)
t
where rn<=2;
```
