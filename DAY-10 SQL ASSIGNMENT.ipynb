```sql
-- ============================================
-- PART A: RANKING FUNCTIONS
-- ============================================
```

```sql
-- Q1. Rank all watches by price (descending) using RANK(). Show model_name, price, and the rank.
select  model_name, price,
RANK()over(
    order by price desc

)AS hr
from watch_sales;
```

```sql
-- Q2. Repeat Q1 using DENSE_RANK(). Add a comment explaining what would differ from RANK() if two watches had the exact same price.

select model_name,price,
DENSE_RANK() OVER(
--If two watches have the same price, DENSE_RANK() gives them the same rank and does not skip the next rank, whereas RANK() gives them the same rank but skips the next rank.
    ORDER BY PRICE DESC
)AS HR
FROM WATCH_SALES;
```

```sql
-- Q3. Assign a unique sequential number to every sale ordered by sale_date (oldest first) using ROW_NUMBER().
select sale_id,brand_name,
ROW_NUMBER()OVER(
    ORDER BY SALE_DATE ASC
)AS USN
FROM WATCH_SALES;
```

```sql
-- Q4. Within each category, rank watches by units_sold (highest first) using RANK(), resetting for every category (use PARTITION BY).

select category,model_name,units_sold,
Rank() over(
    partition by category
    order by  units_sold desc
)as RANK
FROM WATCH_SALES;
```

```sql
-- Q5. Find the top 3 best-selling watches (by units_sold) in each category using DENSE_RANK() inside a CTE or subquery.

select*
from(
    select *,
    DENSE_RANK()OVER (
        PARTITION BY category
        ORDER BY units_sold DESC
    )AS best
    FROM watch_sales
)t
--t is just a temporary name (alias) for the result of your inner query.
where best<=3;
```

```sql
-- Q6. Use ROW_NUMBER() to identify and list only the most recent sale for each brand_name.

select *
from (
    select sale_date,brand_name,model_name,
        Row_Number()over(
        Partition By  trim(brand_name)
        order by sale_date desc
        )AS Recent
    from watch_sales
)
t
where Recent =1;
```

```sql
-- Q8.  Use CONCAT() to build a full_name column from customer_first_name and customer_last_name, separated by a space.

SELECT 
CONCAT(CUSTOMER_FIRST_NAME,' ',CUSTOMER_LAST_NAME )AS FULL_NAME
FROM WATCH_SALES;
```

```sql
-- Q9.  brand_name has extra leading/trailing spaces (e.g. ' Casio '). Use TRIM() to clean it up.

SELECT brand_name,
TRIM(brand_name) as CN
FROM watch_sales;
```

```sql
-- Q10. Extract the first 4 characters of model_name for every row using SUBSTRING().
select model_name,
SUBSTRING(model_name ,1,4)As extracted
from watch_sales;
```

```sql
-- Q11. model_name uses hyphens (e.g. 'Submariner-Date'). Use REPLACE() to convert all hyphens to spaces.
select model_name,
REPLACE(model_name ,'-',' ' )as NEW
from watch_sales;
```

```sql
-- Q12. Standardize the country column to all uppercase using UPPER(). Then write a second version using LOWER().
select country,
upper(country)AS UC,
lower(country)as LC
from watch_sales;
```

```sql
-- Q13. Combine TRIM() and UPPER() in a single query to clean and standardize brand_name in one step.

select brand_name,
upper(trim(brand_name))as New_brand
from watch_sales;
```

```sql
-- ============================================
-- PART C: MIXED / COMBINED PRACTICE
-- ============================================

-- Q14. Clean customer_first_name and customer_last_name with TRIM() and proper case (UPPER/LOWER), combine them into full_name with CONCAT(), then rank customers by total units_sold using DENSE_RANK().

WITH customer_data AS
(
    SELECT
        CONCAT(
            UPPER(SUBSTRING(TRIM(customer_first_name), 1, 1)),
            LOWER(SUBSTRING(TRIM(customer_first_name), 2)),
            ' ',
            UPPER(SUBSTRING(TRIM(customer_last_name), 1, 1)),
            LOWER(SUBSTRING(TRIM(customer_last_name), 2))
        ) AS full_name,
        units_sold
    FROM watch_sales
)
SELECT
    full_name,
    units_sold,
    DENSE_RANK() OVER (ORDER BY units_sold DESC) AS units_rank
FROM customer_data;
```

```sql
-- Q15. Using SUBSTRING() and CONCAT(), mask customer_last_name so only the first letter shows and the rest is asterisks (e.g. 'KUMAR' -> 'K****'). Then use ROW_NUMBER() to number the masked list alphabetically by customer_last_name.

SELECT
    ROW_NUMBER() OVER (
        ORDER BY customer_last_name
    ) AS row_num,

    CONCAT(
        SUBSTRING(customer_last_name, 1, 1),
        REPEAT('*', LENGTH(customer_last_name) - 1)
    ) AS masked_last_name

FROM watch_sales
ORDER BY customer_last_name;



```
