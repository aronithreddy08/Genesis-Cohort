```python
%sql
-- Q1 (Medium): Find the total OilProduced_bbl and total GasProduced_mcf for each Field, but only include 'Active' wells.
-- Concepts: SUM, GROUP BY, WHERE

SELECT
    FieldName,
    SUM(OilProduced_bbl) AS Total_OilProduced_bbl,
    SUM(GasProduced) AS Total_GasProduced
FROM well_production
WHERE Status = 'Active'
GROUP BY FieldName;
```

```python
%sql
-- Q2 (Medium): Find the average OperatingCost for each OperatorName,sorted from highest to lowest average cost.
-- Concepts: AVG, GROUP BY, ORDER BY

SELECT
    OperatorName,
    AVG(OperatingCost) AS Average_OperatingCost
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OperatingCost DESC;
```

```python
%sql
-- Q3 (Medium): Find the number of wells per Region, but only for regions that have more than 2 wells.
-- Concepts: COUNT, GROUP BY, HAVING
SELECT
    Region,
    COUNT(WellID) AS NumberOfWells
FROM well_production
GROUP BY Region
HAVING COUNT(WellID) > 2;
```

```python
%sql
-- Q4 (Medium): Display the top 3 Fields with the highest total OilProduced_bbl(combine oil production across all wells in that field).
-- Concepts: SUM, GROUP BY, ORDER BY, LIMIT
SELECT
    FieldName,
    SUM(OilProduced_bbl) AS Total_OilProduced_bbl
FROM well_production
GROUP BY FieldName
ORDER BY Total_OilProduced_bbl DESC
LIMIT 3;
```

```python
%sql
-- Q5 (Medium): Find the minimum and maximum DepthMeters for each WellType,but only consider wells that are NOT 'Abandoned'.
-- Concepts: MIN, MAX, GROUP BY, WHERE
SELECT
    WellType,
    MIN(DepthMeters) AS Minimum_Depth,
    MAX(DepthMeters) AS Maximum_Depth
FROM well_production
WHERE Status <> 'Abandoned'
GROUP BY WellType;
```

```python
%sql
-- Q6 (Medium): List each unique OperatorName along with the total number of wells and average OilProduced_bbl, sorted by average oil production descending, showing only the top 5.
-- Concepts: COUNT, AVG, GROUP BY, ORDER BY, LIMIT

SELECT
    OperatorName,
    COUNT(WellID) AS Total_Wells,
    AVG(OilProduced_bbl) AS Average_OilProduced_bbl
FROM well_production
GROUP BY OperatorName
ORDER BY Average_OilProduced_bbl DESC
LIMIT 5;
```
