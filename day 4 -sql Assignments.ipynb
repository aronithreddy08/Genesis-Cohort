```sql
-- Q5 ():
-- Display orders where the City is Dallas, Austin, or Chicago.
-- Concepts: IN
select *
from customer_orders 
where city in ('Dallas','Austin','Chicago');
```

```sql
-- Q6 ():
-- Find products whose Price is BETWEEN 300 and 1000.
-- Concepts: BETWEEN
select *
from customer_orders
where price BETWEEN 300 and 1000 ;
```

```sql
-- Q7 ():
-- Find all customers whose name starts with 'A' or ends with 'a'.
-- Concepts: LIKE

select *
from customer_orders
where  CustomerName like 'A%'
OR  CustomerName Like '%a' ;
```

```sql
-- Q8 ():
-- Display all records where the City IS NULL.
-- Concepts: IS NULL
 
select *
from customer_orders
where city IS NULL;
```

```sql
-- Q9 ():
-- Find the total sales amount (Quantity × Price) for each Category.
-- Concepts: SUM, GROUP BY

select Category,
sum(Quantity*Price) as avg_price 
from customer_orders
group by Category;


```

```sql
-- Q10 ():
-- Find the average product price for each SalesPerson, sorted from highest to lowest average price.
-- Concepts: AVG, GROUP BY, ORDER BY

-- SELECT salesperson,
--        AVG(price) AS avg_price
-- FROM customer_orders
-- GROUP BY salesperson
-- ORDER BY avg_price DESC;

select salesperson,
    AVG(Price) AS avg_price
FROM customer_orders
group by salesperson
order by avg_price DESC;


```

```sql
-- Q11 ():
-- Find the number of orders placed in each City but only show cities having more than 3 orders.
-- Concepts: COUNT, GROUP BY, HAVING

select city,
count(*) as order_placed
from Customer_orders
group by city
having order_placed >2;


```

```sql
-- Q12 ():
-- Display the Top 3 Products with the highest total sales amount (Quantity × Price).
-- Concepts: SUM, GROUP BY, ORDER BY, LIMIT

select Product,
sum(Quantity * Price) as total_sales
from customer_orders
group by Product
order by total_sales Desc
Limit 3

```
