```sql
DROP TABLE IF EXISTS orders;
DROP TABLE IF EXISTS customers;
DROP TABLE IF EXISTS products;

CREATE TABLE customers (
    cust_id INT PRIMARY KEY,
    cust_name VARCHAR(100),
    city VARCHAR(50),
    signup_date DATE
);

CREATE TABLE products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    price DECIMAL(10,2)
);

CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    cust_id INT,
    product_id INT,
    quantity INT,
    order_date DATE,
    FOREIGN KEY (cust_id) REFERENCES customers(cust_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);

```

```sql
--Q5. Write a query to create a view named delhi_customers showing all customers from the customers table who live in 'Delhi'.

Create or Replace  VIEW delhi_customers as 
select *
from customers
where city ='Delhi';

SELECT *
FROM delhi_customers;
```

```sql
--Q6. Write a query to create a view named order_details_view that joins orders, customers, and products to display: order_id, cust_name, product_name, quantity, and order_date.

Create Or Replace View order_details_view As
Select 
    o.order_id,
    c.cust_name,
    p.product_name,
    o.quantity,
    o.order_date
from orders o
join Customers c
    on o.cust_id=c.cust_id
join products p
    on o.product_id=p.product_id;

select * from order_details_view;
```

```sql
-- Q7. Write a query to create a view named electronics_orders that shows only orders where the product category is 'Electronics', displaying order_id, cust_name, product_name, and quantity.

Create Or Replace View electronics_orders as
Select 
    o.order_id,
    c.cust_name,
    p.product_name,
    o.quantity,
    p.category
from orders o 
join  Customers c 
    on o.cust_id=c.cust_id
join  products p
    on p.product_id=o.product_id
 
where category ='Electronics';

select * from electronics_orders;
```

```sql
--Q8. Using the view from Q6, write a query to find the total quantity ordered by each customer.

select 
    cust_name,
sum(quantity) as total_quantity
from order_details_view
group by cust_name;
```
