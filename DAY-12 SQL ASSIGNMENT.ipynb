```sql
-- Q3. CASE WHEN
-- Using the sales table, create a column called Sales_Category:
--
-- Sales_Amount >= 1500  → 'High'
-- Sales_Amount >= 800   → 'Medium'
-- Otherwise             → 'Low'

Select 
    sale_id,
    customer_name,
    sales_amount,
Case
    when sales_amount <=700 then 'low_range'
    when sales_amount between 701 and 1000 then 'mid_range'
    else 'high_range'
    end as Sales_category
from sales;

```

```sql
-- Q4. CASE WHEN
-- Create a column called Order_Result:
--
-- Status = 'Completed' → 'Successful'
-- Status = 'Cancelled' → 'Unsuccessful'
--
-- Display Sale_ID, Customer_Name, Status and Order_Result.

select 
    sale_id,
    customer_name,
    status,
    Case
        When status ='Completed' then 'Successful'
        when status ='Cancelled' then 'Unsuccessful'
    End as Order_Result
from sales;
```

```sql
-- Q5. LAG()
-- For each customer, display:
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Previous_Month_Sales
--
-- Use LAG() to get the previous month's sales.

select 
    customer_name,
    sale_date,
    sales_amount,
LAG(sales_amount) over(
    partition by customer_name
    order by sale_date
) as previous_sales
from sales;
```

```sql
-- Q6. LEAD()
-- For each customer, display:
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Next_Month_Sales
--
-- Use LEAD() to get the next month's sales.

select 
    Customer_Name,
    Sale_Date,
    Sales_Amount,
LEAD(Sales_Amount) over(
    partition by customer_name
    order by Sale_date
)as Next_Month_Sales
from sales;
```

```sql
-- Q7. MONTH-TO-MONTH SALES CHANGE
-- Calculate the difference between the current month's
-- sales and the previous month's sales.
--
-- Formula:
-- Current Sales - Previous Month Sales
--
-- Use LAG().
with sales_lag as (
select 
    Customer_Name,
    Sale_Date,
    Sales_Amount,    
LAG(Sales_Amount)over(
    partition by customer_name
    order by sale_date
)as previous_month_sales
from sales
)
select 
    customer_name,
    sale_date,
    sales_amount,
    Sales_amount-Previous_month_sales as sales_change
from sales_lag;
```

```sql
-- Q8. MONTH-TO-MONTH PERCENTAGE
-- Calculate the month-to-month percentage change in sales.
--
-- Formula:
-- (Current Sales - Previous Sales)
-- / Previous Sales * 100
--
-- Use LAG().

with sales_lag as(
    select 
    Customer_Name,
    Sale_Date,
    Sales_Amount, 
    LAG(Sales_Amount) over(
        partition by customer_name
        order by sale_date
    )as previous_sales

from sales
)
select 
    Customer_Name,
    Sale_Date,
    Sales_Amount, 
    previous_sales,
    round(
        (Sales_Amount- Previous_Sales)/ Previous_Sales * 100,2
    ) As Month_percentage
from sales_lag;
```

```sql
-- Q9. SALES TREND
-- Using LAG() and CASE WHEN, create a column called Sales_Trend:
--
-- No previous sale → 'First Month'
-- Current Sales > Previous Sales → 'Increased'
-- Current Sales < Previous Sales → 'Decreased'
-- Current Sales = Previous Sales → 'No Change'

with sales_lag AS (
    select 
        customer_name,
        sales_amount,
        sale_date,
    LAG(sales_amount)over(
        partition by customer_name
        order by sale_date
    )as previous_sales
    from sales
)
select
        customer_name,
        sales_amount,
        sale_date,
        previous_sales,
        
        CASE 
        when  sales_amount < previous_sales then 'Decreased'
        when  sales_amount > previous_sales then 'increased'
        when  sales_amount =previous_sales then 'No change'
         end as sales_Trend
from sales_lag;


```

```sql
-- Q10. LAG() + LEAD()
-- Display the following for each customer:
--
-- Customer_Name
-- Sale_Date
-- Sales_Amount
-- Previous_Month_Sales
-- Next_Month_Sales
--
-- Use both LAG() and LEAD() in the same query.

SELECT
    Customer_Name,
    Sale_Date,
    Sales_Amount,
    LAG(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Previous_Month_Sales,
    LEAD(Sales_Amount) OVER (
        PARTITION BY Customer_Name
        ORDER BY Sale_Date
    ) AS Next_Month_Sales
FROM sales;
     
```
