```sql
    --1. CURRENT_DATE
--Display customer_name, order_date, and today's date.

select customer_name, order_date,
current_date() AS today_date
from audio_sales;
```

```sql
--2. DATEDIFF
--Calculate the number of days between order_date and delivery_date.

select order_date ,delivery_date,
DATEDIFF(delivery_date,order_date )as NOD
from audio_sales;
```

```sql
-- 3. DATE_ADD
-- Find the date that is 7 days after each order_date.

select order_date,
DATE_ADD(order_date,7)as Date_After_7_DAYS
from audio_sales;
```

```sql
-- 4) Extract the year, month, and day from order_date.\

select order_date,
EXTRACT(YEAR from order_date) AS Orderd_Year,
EXTRACT(MONTH from order_date) AS Orderd_Month,
EXTRACT(YEAR from order_date) AS Orderd_Days
from audio_sales;
```

```sql
-- 5. CASE WHEN
-- Create product_type:

select 
    product_name,
    price,
CASE 
    WHEN price <100 Then 'Budget'
    WHEN price Between 101 And 201 Then 'Mid_Range'
    Else 'Premium'
End as Product_type
from audio_sales;
```

```sql
-- 6.CASE WHEN
-- Create product_type:
-- AirPods → Apple Product

Select
    product_name,
    category,
CASE
    When category ='AirPods' Then 'Apple Product'
    When category ='Earphones' Then  'Other Brand'
End as Product_type
from audio_sales;
```

```sql
-- 7. COALESCE
-- Replace NULL values in discount with 0.

Select 
    customer_name,
    discount,
COALESCE(discount,0)AS Final_discount
from audio_sales;

    
```

```sql
-- 8. COALESCE
-- Replace NULL values in rating with 0.
Select 
    product_name,
    rating,
COALESCE(rating,0) AS Final_Rating
from audio_sales;
```

```sql
-- 9. NULLIF
-- Use NULLIF to convert a rating of 0 into NULL.

Select 
    product_name,
    rating,
NULLIF(rating,0)AS rating_with_zeros
from audio_sales;
```

```sql
-- 10. NULLIF
-- Calculate discount percentage using:
-- discount / price × 100
-- Use NULLIF to prevent division-by-zero.

Select
    price,
    discount,
(discount/nullif(price,0)) *100 AS Discount_perc
from audio_sales;

```

```sql
-- 11. Combined
-- Find the delivery days and create a category:
-- 3 days or less → Fast
-- 4–5 days → Normal
-- More than 5 days → Slow

Select 
    customer_name,
    order_date,
    delivery_date,
DATEDIFF(delivery_date,order_date) AS Nof_days,
CASE 
    When DATEDIFF(order_date,delivery_date)<=3
     Then 'Fast'
    When DATEDIFF(order_date,delivery_date )between 4 And 5 
    Then 'Normal'
    Else 'slow'
end as speed
from audio_sales;
```

```sql
-- 12. Combined
-- Display the final discount using COALESCE, calculate sales using price × quantity, and classify sales using CASE WHEN.
select 
    price,
    discount,
    quantity,
COALESCE(discount/nullif(price,0)*100,0) as FD,
(price*quantity) as sales,
Case 
    When (price*quantity)<200
    Then 'low sales'
    When  (price*quantity) between 201 And 301
    Then 'Mid_range'
    Else 'Top Sales'
End As Sale_Category
from audio_sales;
```
