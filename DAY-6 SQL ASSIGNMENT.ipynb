```sql
-- CREATE TABLES & INSERT DATA
-- TABLE 1 : CUSTOMERS

CREATE OR REPLACE TABLE Customers (
    customer_id INT,
    customer_name STRING,
    city STRING
);

-- Insert customer records
INSERT INTO Customers VALUES
(1,'John','Dallas'),
(2,'Emma','Austin'),
(3,'Michael','Houston'),
(4,'Sophia','Dallas'),
(5,'James','Austin');

-- View the table
SELECT * FROM Customers;
```

```sql

-- TABLE 2 : AGENTS

CREATE OR REPLACE TABLE Agents (
    agent_id INT,
    agent_name STRING,
    office_city STRING
);

-- Insert agent records
INSERT INTO Agents VALUES
(101,'David','Dallas'),
(102,'Sarah','Austin'),
(103,'Robert','Houston');

-- View the table
SELECT * FROM Agents;
```

```sql
-- TABLE 3 : PROPERTIES

CREATE OR REPLACE TABLE Properties (
    property_id INT,
    property_name STRING,
    city STRING,
    property_type STRING,
    price INT,
    agent_id INT
);

-- Insert property records
INSERT INTO Properties VALUES
(1001,'Green Villa','Dallas','Villa',450000,101),
(1002,'Lake House','Austin','House',620000,102),
(1003,'Sky Apartments','Houston','Apartment',380000,103),
(1004,'Oak Residency','Dallas','House',520000,101),
(1005,'River View','Austin','Villa',700000,102),
(1006,'Palm Residency','Houston','House',490000,103);

-- View the table
SELECT * FROM Properties;
```

```sql

-- TABLE 4 : SALES

CREATE OR REPLACE TABLE Sales (
    sale_id INT,
    customer_id INT,
    property_id INT,
    sale_date DATE
);

-- Insert sales records
INSERT INTO Sales VALUES
(5001,1,1001,'2026-01-10'),
(5002,2,1002,'2026-01-18'),
(5003,3,1003,'2026-02-05'),
(5004,4,1004,'2026-03-11'),
(5005,1,1005,'2026-04-20'),
(5006,5,1006,'2026-05-15');

-- View the table
SELECT * FROM Sales;
```

```sql
--1. Display customer name, property purchased, price, and sale date

SELECT c.customer_name,
       p.property_name,
       p.price,
       s.sale_date
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id;
```

```sql
--2. Display property details with the managing agent

SELECT p.property_name,
       p.city,
       p.property_type,
       p.price,
       a.agent_name
FROM Properties p
INNER JOIN Agents a
ON p.agent_id = a.agent_id;
```

```sql
--3. Display customer, property, and agent details for each sale

SELECT c.customer_name,
       p.property_name,
       a.agent_name
FROM Sales s
INNER JOIN Customers c
ON s.customer_id = c.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
INNER JOIN Agents a
ON p.agent_id = a.agent_id;
```

```sql
--4. Find customers who bought properties costing more than 500000

SELECT c.customer_name,
       p.property_name,
       p.price
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
WHERE p.price > 500000;
```

```sql
--5.Display properties managed by Sarah

SELECT p.property_name,
       p.city,
       p.price,
       a.agent_name
FROM Properties p
INNER JOIN Agents a
ON p.agent_id = a.agent_id
WHERE a.agent_name = 'Sarah';
```

```sql
--6. Find customers who purchased properties in their own city

SELECT c.customer_name,
       c.city,
       p.property_name,
       p.city AS property_city
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
WHERE c.city = p.city;
```

```sql
-- 7.Calculate total sales amount handled by each agent

SELECT a.agent_name,
       SUM(p.price) AS total_sales
FROM Agents a
INNER JOIN Properties p
ON a.agent_id = p.agent_id
INNER JOIN Sales s
ON p.property_id = s.property_id
GROUP BY a.agent_name;
```

```sql
--8. Count properties purchased by each customer

SELECT c.customer_name,
       COUNT(s.property_id) AS properties_purchased
FROM Customers c
INNER JOIN Sales s
ON c.customer_id = s.customer_id
GROUP BY c.customer_name;
```

```sql
--9. Find the highest-priced property sold by each agent

SELECT a.agent_name,
       MAX(p.price) AS highest_property_price
FROM Agents a
INNER JOIN Properties p
ON a.agent_id = p.agent_id
INNER JOIN Sales s
ON p.property_id = s.property_id
GROUP BY a.agent_name;
```

```sql
--10. Complete sales report with customer, property, agent, and sale details

SELECT s.sale_id,
       c.customer_name,
       c.city AS customer_city,
       p.property_name,
       p.property_type,
       p.price,
       a.agent_name,
       a.office_city,
       s.sale_date
FROM Sales s
INNER JOIN Customers c
ON s.customer_id = c.customer_id
INNER JOIN Properties p
ON s.property_id = p.property_id
INNER JOIN Agents a
ON p.agent_id = a.agent_id
ORDER BY s.sale_date;
```
